### Bài 7. Các hàm thống kê trên Tensor
**Yêu cầu:** Thay dim=0 bằng dim=1 và nhận xét sự khác nhau.

```python
import torch
a = torch.randn(3, 4)
print(a)

print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))

print("\n--- Khi thay bằng dim=1 ---")
print("\nMean (dim=1):\n", torch.mean(a, dim=1))
print("\nStandard deviation (dim=1):\n", torch.std(a, dim=1))
print("\nCumulative sum (dim=1):\n", torch.cumsum(a, dim=1))
```

**Trả lời Bài 7:**

- Khi thay bằng `dim=1`, các hàm thống kê được tính dọc theo các hàng (từ trái qua phải của mỗi hàng).

### Bài 8. Autograd cơ bản
**Yêu cầu:**
- Tính thủ công đạo hàm của y = x^2 tại x = 3.6 và so sánh với x.grad.
- Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

```python
import torch

x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

x_new = torch.tensor(3.6, requires_grad=True)
y_new = 3 * x_new**2 + 2 * x_new + 1
y_new.backward()

print("\n--- Với hàm y = 3*x**2 + 2*x + 1 ---")
print("x =", x_new)
print("y =", y_new)
print("x.grad =", x_new.grad)
```

**Trả lời Bài 8:**
- Tính thủ công hàm $y = x^2$: $y' = 2x$. Tại $x = 3.6 \Rightarrow y' = 2 \times 3.6 = 7.2$. Kết quả khớp hoàn toàn với `x.grad`.
- Tính thủ công hàm $y = 3x^2 + 2x + 1$: $y' = 6x + 2$. Tại $x = 3.6 \Rightarrow y' = 6 \times 3.6 + 2 = 23.6$. Kết quả in ra của `x_new.grad` bằng 23.6, hoàn toàn chính xác.

### Bài 9. Gradient Descent tìm hệ số đa thức
**Yêu cầu:**
- So sánh ba hệ số học được với [1, 2, 3].
- Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([X * X, X, np.ones_like(X)])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)
```

**Trả lời Bài 9:**
- Ba hệ số học được đã hội tụ sát về giá trị [1, 2, 3] tương ứng với các hệ số của đa thức.
- `optimizer.zero_grad()`: Reset toàn bộ gradient của bước cập nhật trước đó về 0 để tránh bị cộng dồn.
- `mse.backward()`: Kích hoạt autograd, tính toán đạo hàm (gradient) của hàm loss (mse) theo từng trọng số (w).
- `optimizer.step()`: Cập nhật các trọng số theo hướng ngược lại của gradient để cực tiểu hóa sai số.

### Bài 10. Giải hệ phương trình bằng Autograd
**Yêu cầu:**
- Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.
- Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)

print("\n--- Kiểm tra nghiệm ---")
print(f"A + B = {A.item() + B.item():.4f} (Kỳ vọng: 9)")
print(f"C - D = {C.item() - D.item():.4f} (Kỳ vọng: 1)")
print(f"A + C = {A.item() + C.item():.4f} (Kỳ vọng: 8)")
print(f"B - D = {B.item() - D.item():.4f} (Kỳ vọng: 2)")
```

**Trả lời Bài 10:**
- Bài toán có vô số nghiệm vì trong hệ 4 phương trình này có sự phụ thuộc tuyến tính lẫn nhau. Cụ thể: (A + B) - (B - D) = A + D = 7, và (A + C) - (C - D) = A + D = 7. Nên thực chất hệ chỉ tương đương với 3 phương trình độc lập cho 4 ẩn.

### Bài 11. Bài tập áp dụng
**Yêu cầu:**
- Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4x6, 2x12 và 2x3x4.
- Tạo hai Tensor A, B kích thước 3x4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.
- Cho y = 2x^2 + 5x + 3. Dùng autograd tính dy/dx tại x=2 và kiểm tra bằng phép tính thủ công.
- Thay đa thức trong Bài 9 bằng y = 2x^2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import torch
import numpy as np

# Yêu cầu 1
X = torch.arange(1, 25)
print("Tensor X (4x6):\n", X.reshape(4, 6))
print("\nTensor X (2x12):\n", X.reshape(2, 12))
print("\nTensor X (2x3x4):\n", X.reshape(2, 3, 4))

# Yêu cầu 2
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("\nA + B =\n", A + B)
print("\nA * B =\n", A * B)
print("\nA / B =\n", A / B)
print("\nTrung bình theo cột:\n", torch.mean(A, dim=0))
print("\nĐộ lệch chuẩn theo cột:\n", torch.std(A, dim=0))

# Yêu cầu 3
x = torch.tensor(2.0, requires_grad=True)
y = 2*x**2 + 5*x + 3
y.backward()
print("\ndy/dx tại x=2:", x.grad.item())
print("Tính thủ công: y' = 4x + 5. Tại x=2, y' = 4*2 + 5 = 13. (Khớp)")

# Yêu cầu 4
poly = np.poly1d([2, -3, 5])
X_dt = np.random.randn(20, 1) * 5
Y_dt = poly(X_dt)
XX_dt = np.hstack([X_dt * X_dt, X_dt, np.ones_like(X_dt)])

w_new = torch.randn(3, 1, requires_grad=True)
x_new = torch.tensor(XX_dt, dtype=torch.float32)
y_new = torch.tensor(Y_dt, dtype=torch.float32)

opt_new = torch.optim.NAdam([w_new], lr=0.1)

for _ in range(2000):
    opt_new.zero_grad()
    pred = x_new @ w_new
    loss = torch.mean(torch.square(y_new - pred))
    loss.backward()
    opt_new.step()

print("\nHệ số học được cho y = 2x^2 - 3x + 5:\n", w_new.data)
```
