### Bài 1. Tạo Tensor cơ bản
**Yêu cầu:** Cho biết kích thước, số chiều và kiểu dữ liệu của Tensor a.

```python
import torch
a = torch.tensor(
    [[1, 2, 3], 
    [4, 5, 6]], 
    dtype=torch.int32
    )

print('Tensor a:\n', a)
print('Kích thước:', a.shape)
print('Số chiều:', a.ndim)
print('Kiểu dữ liệu:', a.dtype)
```

**Trả lời:**
- Kích thước: `(2, 3)`
- Số chiều: `2`
- Kiểu dữ liệu: `torch.int32`

### Bài 2. Tạo các Tensor thường dùng
**Yêu cầu:**
- Nhận xét sự khác nhau giữa torch.rand() và torch.randn().
- Tạo thêm một Tensor 5x5 chứa toàn giá trị 10.

```python
import torch

a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print(a)
print(b)
print(c)
print(d)
print(e)

f = torch.full((5, 5), 10)
print('Tensor 5x5 toàn 10:\n', f)
```

**Trả lời:**
- `torch.rand()`: Sinh số ngẫu nhiên theo phân phối đều trong khoảng [0, 1).
- `torch.randn()`: Sinh số ngẫu nhiên theo phân phối chuẩn tắc (mean = 0, std = 1).

### Bài 3. Kiểm tra thông tin Tensor
**Yêu cầu:** Giải thích ý nghĩa của torch.Size([2, 3, 4]).

```python
import torch
a = torch.zeros(2, 3, 4)
print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

**Trả lời:**
- `torch.Size([2, 3, 4])` nghĩa là tensor 3 chiều. Chiều 0 có 2 phần tử, chiều 1 có 3 phần tử, chiều 2 có 4 phần tử. (Tổng có 2x3x4 = 24 phần tử).

### Bài 4. Slicing và Indexing
**Yêu cầu:**
- Cho biết shape của a[1] và a[1:, 2:4].
- Thử lấy hai phần tử cuối cùng trên chiều thứ ba.

```python
import torch
a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])

phan_tu_cuoi = a[:, :, -2:]
print("\nHai phần tử cuối cùng trên chiều thứ ba:\n", phan_tu_cuoi)
print("Shape của hai phần tử cuối:", phan_tu_cuoi.shape)
```

**Trả lời:**
- Shape của `a[1]` là `torch.Size([4, 5])`.
- Shape của `a[1:, 2:4]` là `torch.Size([2, 2, 5])`.
- Code lấy 2 phần tử cuối: `a[:, :, -2:]`.

### Bài 5. Reshape và Flatten
**Yêu cầu:**
- Kiểm tra số phần tử trước và sau khi reshape.
- Thử reshape Tensor thành kích thước 2x6.

```python
import torch
a = torch.randn(3, 4)

print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))

print("\nSố phần tử trước khi reshape:", a.numel())
print("Số phần tử sau khi reshape:", a.reshape(3, 2, 2).numel())

print("\nReshape thành kích thước 2x6:\n", a.reshape(2, 6))
```

**Trả lời Bài 5:**
- Số phần tử trước và sau khi reshape không thay đổi (đều bằng 12 phần tử).
